## 🌿 Final Step: Run EcoSim on All Gene Trees and Sequences

Now that we have:
- `final_fastas/` → trimmed, cleaned FASTA files (with aligned outgroups),
- `rerooted_trees/` → rerooted Newick trees for each gene,

We can run **EcoSim** in batch mode to generate XML results for each gene.

### ✅ What this script does:
- Loops through each FASTA file in `final_fastas/`
- Locates the corresponding rerooted tree in `rerooted_trees/`
- Runs the EcoSim executable on both files
- Saves the XML output to `ecosim_results/`

> ⚠️ Ensure that your EcoSim binary path (e.g., `C:/ecosim/bin/ecosim.exe`) is correctly specified in the script below.

```python
import os
import subprocess
```

```python
def run_ecosim_batch(fasta_dir, tree_dir, output_dir,
                     ecosim_jar="C:/ecosim/ecosim.jar",
                     ecosim_dir="C:/ecosim"):
    os.makedirs(output_dir, exist_ok=True)

    for fasta_file in os.listdir(fasta_dir):
        if not fasta_file.endswith(".fasta"):
            continue

        gene_name = fasta_file.replace(".fasta", "")
        fasta_path = os.path.abspath(os.path.join(fasta_dir, fasta_file))
        tree_path = os.path.abspath(os.path.join(tree_dir, f"{gene_name}_tree.nwk"))
        xml_output = os.path.abspath(os.path.join(output_dir, f"{gene_name}_results.xml"))

        if not os.path.exists(tree_path):
            print(f"⚠️ Missing tree for {gene_name}, skipping.")
            continue

        print(f"🌱 Running EcoSim for {gene_name}...")
        try:
            subprocess.run([
                "java", "-Xmx12G", "-jar", ecosim_jar,
                f"-s={fasta_path}",
                f"-p={tree_path}",
                f"-o={xml_output}",
                "-d", "-r", "-n"
            ], check=True, cwd=ecosim_dir)
            print(f"✅ EcoSim output saved: {xml_output}")
        except subprocess.CalledProcessError as e:
            print(f"❌ EcoSim error for {gene_name}: {e}")
```

```python
# Run the pipeline:
run_ecosim_batch("tree_rdy_fastas", "rerooted_trees", "ecosim_results")
```
